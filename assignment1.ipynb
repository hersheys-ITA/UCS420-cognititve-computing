```python
n = int(input("Enter a Number: "))

sum = 0

for num in range(2, n + 1):
    prime = True

    for i in range(2, num // 2 + 1):
        if num % i == 0:
            prime = False
            break

    if prime:
        sum = sum + num

print("Sum of Prime Numbers =", sum)
```

```python
def IsPrime(num):
    if num < 2:
        return False

    for i in range(2, num // 2 + 1):
        if num % i == 0:
            return False

    return True


def AddPrime(n):
    s = 0
    for i in range(2, n + 1):
        if IsPrime(i):
            s = s + i
    return s


n = int(input("Enter a Number: "))
print("Sum of Prime Numbers -->", AddPrime(n))
```
