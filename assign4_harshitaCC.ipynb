#question 2

```python
import pandas as pd

data = {
    "question": [
        "How do I reset my password?",
        "How can I change my password?",
        "How do I contact support?",
        "How do I reset my account?"
    ],
    "category": [
        "Account",
        "Account",
        "Support",
        "Account"
    ],
    "keywords": [
        "password, reset",
        "password, change",
        "support, contact",
        "account, reset"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
#question 2
def score_hypothesis(query):
    query = query.lower()
    results = []

    for index, row in df.iterrows():
        text = str(row['question']).lower()
        keywords = str(row['keywords']).lower()

        score = 0

        if query in text:
            score += 2

        if query in keywords:
            score += 3

        if score > 0:
            results.append((index, row['question'], score))

    results.sort(key=lambda x: x[2], reverse=True)

    return results


# Example
print(score_hypothesis("password"))
```

```python
#question 6
def score_hypothesis(query):
    query = query.lower()
    results = []

    for index, row in df.iterrows():
        text = str(row["question"]).lower()
        keywords = str(row["keywords"]).lower()

        score = 0

        if query in text:
            score += 2

        if query in keywords:
            score += 3

        if score > 0:
            results.append((index, row["question"], score))

    if not results:
        return []

    highest_score = max(x[2] for x in results)

    return [x for x in results if x[2] == highest_score]


print(score_hypothesis("password"))
```
