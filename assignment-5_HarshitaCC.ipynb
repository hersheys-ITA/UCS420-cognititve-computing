```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed array:", arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
x_values, x_counts = np.unique(x, return_counts=True)
x_max = np.max(x_counts)
x_frequent = x_values[x_counts == x_max]

print("Most frequent value(s) in x:", x_frequent)
for value in x_frequent:
    print("Indices:", np.where(x == value)[0])

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
y_values, y_counts = np.unique(y, return_counts=True)
y_max = np.max(y_counts)
y_frequent = y_values[y_counts == y_max]

print("Most frequent value(s) in y:", y_frequent)
for value in y_frequent:
    print("Indices:", np.where(y == value)[0])
```
